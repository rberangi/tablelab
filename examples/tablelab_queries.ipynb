# TableLab queries

A GUI for selecting, joining, filtering, summarising, and cleaning tables, with
reproducible code export to pandas, PySpark, Spark SQL, MySQL, PostgreSQL, SQLite,
and DuckDB.

Run the launcher below. In a source checkout it uses the local `src/tablelab` package,
whether Jupyter starts in the project folder or `examples`. Otherwise install TableLab
in this notebook kernel first with `%pip install tablelab`.

For local development, install the checkout and its dependencies with
`%pip install -e /full/path/to/tablelab` (replace the path). Use `%pip` rather than
`!pip` so installation targets this kernel. Restart the kernel after installing or
updating TableLab, then run the launcher again. SQL database exports require
`sqlglot`, included in the current package dependencies.

The launcher prints the version and package location so you can confirm which
copy is running.

```python
import sys
from pathlib import Path

# Prefer this checkout when running from the project or examples folder.
source = next((p for p in (Path.cwd() / "src", Path.cwd().parent / "src")
               if (p / "tablelab" / "__init__.py").is_file()), None)
if source is not None and str(source) not in sys.path:
    sys.path.insert(0, str(source))

import tablelab
from tablelab import query_main

print("Python:", sys.executable)
print("TableLab:", tablelab.__version__, tablelab.__file__)
app = query_main()
```

## Using the app

- **Load tables** with **Browse…**, a path or URL and **Add table**, or **Upload**. Choose the
  **Source table**, then an operation tab: Select, Join, Filter, Group & summarise, or
  Missing values.
- **Apply and preview result** shows the result a page at a time; the full result is
  `app.result`. **Add as new table, use name:** makes it a source for another query.
  Inputs stay unchanged.
- **View source table** browses the source rows a page at a time. Start row and End row
  set the page; Previous/Next keep its size. Only the visible page is rendered.
- **Inspect a column** shows charts, summary statistics, and categorical value counts
  without changing your query.
- **Add step** saves the current operation to a pipeline so later steps run on its result.
  Pick a **Code output** language and **Show code** for a standalone script of the pipeline.

## Optional sample tables

Run the next cell to add Customers and Orders, or load your own files instead.

- **Join:** select Customers as Source and Orders as Right table. Select `customer_id` on
  both sides, choose Left, then preview. The result includes customers with no orders and
  multiple rows for customers with several orders. For multiple keys, Ctrl/Cmd-click matching
  columns; pairs follow column order. Missing keys match each other, as in pandas.
- **Filter:** select Orders, choose the Filter tab, set `amount` greater than `20`, and preview.
- **Group & summarise:** select Orders, group by `customer_id`, choose Sum and column `amount`,
  then preview. Count rows requires no value column.

Chain operations either with **Add step** or by adding a result as a new table. Changing query
settings clears the previous preview. All loaded tables and results live in memory for this
notebook session.

```python
import pandas as pd

app.add_table(pd.DataFrame({
    "customer_id": [1, 2, 3],
    "name": ["Ada", "Ben", "Cy"],
    "city": ["Sydney", "Melbourne", "Sydney"],
}), "Customers")
app.add_table(pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "customer_id": [1, 1, 2, 4],
    "amount": [10.0, 25.0, 30.0, 40.0],
}), "Orders")
```

## Using results in code

`app.df` is the selected source table. `app.result` is the full previewed result, or `None`
after settings change. To save it from a new code cell, use
`app.result.to_csv("query_result.csv", index=False)`.

To chart a row as a signal (for example ECG beats or spectra) rather than query tables, use the
chart app instead: `from tablelab import main; main()`.
