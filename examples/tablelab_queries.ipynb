Install TableLab in this notebook kernel first (`%pip install tablelab`

```python
from tablelab import query_main

app = query_main()
```

Optional small demo tables for trying joins and queries.


```python
import pandas as pd

app.add_table(pd.DataFrame({
    "customer_id": [1, 2, 3],
    "name": ["Ada", "Ben", "Cy"],
    "city": ["Sydney", "Melbourne", "Sydney"],
}), "Customers")
app.add_table(pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "customer_id": [1, 1, 2, 4],
    "amount": [10.0, 25.0, 30.0, 40.0],
}), "Orders")
```
